#### Ejercicio 15

¿Cuál es la casa más cara de cada población del cinturón sur de Madríd? (★★☆)

> Debes usar el subset obtenido en la pregunta 13

- Genera un DataFrame con esta información
- Muestra tanto la dirección como el precio de la casa seleccionada de cada población.
- Genera conclusiones en una celda markdown

````python
fuenlabrada = south_belt[south_belt['level5'] == 'Fuenlabrada']
leganes = south_belt[south_belt['level5'] == 'Leganés']
getafe = south_belt[south_belt['level5'] == 'Getafe']
alcorcon = south_belt[south_belt['level5'] == 'Alcorcón']

max_fuenlabrada = fuenlabrada[fuenlabrada['price'] == fuenlabrada['price'].max()]
max_leganes = leganes[leganes['price'] == leganes['price'].max()]
max_getafe = getafe[getafe['price'] == getafe['price'].max()]
max_alcorcon = alcorcon[alcorcon['price'] == alcorcon['price'].max()]
````

````python
most_expensive = pd.DataFrame({
    'Población': ['Fuenlabrada', 'Leganés', 'Getafe', 'Alcorcón'],
    'Dirección': [
        max_fuenlabrada['address'].iloc[0],
        max_leganes['address'].iloc[0],
        max_getafe['address'].iloc[0],
        max_alcorcon['address'].iloc[0]
    ],
    'Precio': [
        max_fuenlabrada['price'].iloc[0],
        max_leganes['price'].iloc[0],
        max_getafe['price'].iloc[0],
        max_alcorcon['price'].iloc[0]
    ]
})

most_expensive
````

Entre las cuatro poblaciones analizadas, la vivienda más cara se encuentra en Getafe, con un precio de 1.050.000 €, seguida por Alcorcón con 950.000 €. Por otro lado, las viviendas más caras de Leganés y Fuenlabrada tienen precios de 650.000 € y 490.000 €, respectivamente.

#### Ejercicio 16

¿Qué puedes decir acerca del precio por metro cuadrado (precio/m2) entre los municipios de 'Getafe' y 'Alcorcón'? (★★☆)

> Debes usar el subset obtenido en la pregunta 13
>
> Pista: Crea una nueva columna llamada `pps` (price per square en inglés) y luego analiza los valores

````python
getafe = south_belt[south_belt['level5'] == 'Getafe']
alcorcon = south_belt[south_belt['level5'] == 'Alcorcón']

mean_pps_getafe = getafe['pps'].mean()
mean_pps_alcorcon = alcorcon['pps'].mean()

print(f'Precio promedio por m2 de Getafe: {mean_pps_getafe} €')
print(f'Precio promedio por m2 de Alcorcón: {mean_pps_alcorcon} €')
````

Al comparar el precio promedio por metro cuadrado de Getafe y Alcorcón, se observa que Alcorcón presenta un valor superior. El precio promedio por m² es de aproximadamente 2.242,94 € en Alcorcón y 2.066,31 € en Getafe.
